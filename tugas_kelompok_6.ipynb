```python
class FoodKnowledgeBase:
    def __init__(self):
        # Basis pengetahuan yang dipetakan dari diagram
        self.knowledge = {
            "nasi": "karbohidrat",
            "gandum": "karbohidrat",
            "roti": "karbohidrat",
            "ayam": "protein",
            "telur": "protein"
        }

    def check_nutrition(self, food_name):
        """Mengecek kandungan nutrisi berdasarkan nama makanan."""
        food_lower = food_name.lower()
        if food_lower in self.knowledge:
            result = self.knowledge[food_lower]
            return f"Makanan '{food_name}' mengandung {result}."
        else:
            return f"Maaf, '{food_name}' tidak ditemukan dalam knowledge base."

    def get_all_by_category(self, category):
        """Mencari semua makanan berdasarkan kategori nutrisi."""
        cat_lower = category.lower()
        items = [food for food, nut in self.knowledge.items() if nut == cat_lower]

        if items:
            return f"Daftar makanan sumber {category}: {', '.join(items)}."
        else:
            return f"Kategori {category} tidak ditemukan."

# --- Contoh Penggunaan ---
kb = FoodKnowledgeBase()

# 1. Cek nutrisi spesifik
print(kb.check_nutrition("Ayam"))
print(kb.check_nutrition("Roti"))

# 2. Cek daftar berdasarkan kategori
print(kb.get_all_by_category("karbohidrat"))
print(kb.get_all_by_category("protein"))
```
